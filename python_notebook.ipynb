1. How an image is stored, what are the value range (0, 255), what are the values 
for Red, Green, Blue, Black & White.

An image is stored as a grid of pixels, where each pixel contains color information(RGB). Each channel value ranges from 0 to 255.

- **Red:** (255, 0, 0)
- **Green:** (0, 255, 0)
- **Blue:** (0, 0, 255)
- **Black:** (0, 0, 0)
- **White:** (255, 255, 255)


2. Read an image & store it in different ways (RGB, HSV etc…)

```python

import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\HP\basic_computer_vision\batman image.jpg") 


```

```python
plt.imshow(img)
```

RGB IMAGE

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
```

Gray image

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGRA2GRAY)
plt.imshow(img_rgb)
```

HSV Image

```python
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
plt.imshow(img_hsv)
```

3. From agiven RGB image, split the channels and show each channel seperately

```python
plt.title("Red_image")
plt.imshow(img[:,:,0], cmap="grey")
```

```python
plt.title("Green_image")
plt.imshow(img[:,:,1])
```

```python
plt.title("Blue_image")
plt.imshow(img[:,:,2],cmap="grey")
```
